# PySpark initialization script for Livy sessions

This script fixes the missing PythonSQLUtils import in the Py4J gateway


```pyspark
%%spark

from py4j.java_gateway import java_import  # type: ignore

# Import PythonSQLUtils and other SQL utilities into the gateway namespace
# This fixes the "JavaPackage object is not callable" error when using
# spark.createDataFrame() and other SQL operations

java_import(spark._jvm, "org.apache.spark.sql.api.python.*")  # type: ignore
```
